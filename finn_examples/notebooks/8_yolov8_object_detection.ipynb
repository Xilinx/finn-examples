```python
import cv2
from finn_examples import models
```

# Initialize the Accelerator

```python
# The yolov8 is pre-build only for alveo u55c at the moment
accel = models.yolov8n_4w4a_coco()
```

# Run Inference

```python
img = cv2.imread('images/test.jpg')
detections = accel.single_inference(img, conf_thres=0.3) 
```

# Visualize output

```python
visualized_img = accel.visualize(img, detections)
cv2.imwrite('demo.jpg', visualized_img)
```
